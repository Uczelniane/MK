```python
import numpy as np

LICZBA_SEKWENCJI = 30
KROKI_CZASOWE = 20
MOZLIWE_WARTOSCI = np.array([0.33, 0.66, 1.0])

# Ustawienie ziarna losowości dla powtarzalności wyników
np.random.seed(42)

# 2. Generowanie sekwencji wejściowych (X)
# z MOZLIWE_WARTOSCI, a następnie używamy tych indeksów do utworzenia macierzy X.

# Losowe indeksy z zakresu [0, 2] o kształcie (30, 20)
losowe_indeksy = np.random.randint(0, len(MOZLIWE_WARTOSCI), size=(LICZBA_SEKWENCJI, KROKI_CZASOWE))

# Użycie indeksów do wyboru wartości z MOZLIWE_WARTOSCI
X = MOZLIWE_WARTOSCI[losowe_indeksy]

# 3. Obliczanie celów wyjściowych (t)
t = np.mean(X, axis=1).reshape(-1, 1)

# 4. Weryfikacja i podsumowanie
print(f"Kształt danych wejściowych X (sekwencje, kroki_czasowe): {X.shape}")
print(f"Kształt celów wyjściowych t (sekwencje, 1): {t.shape}")
print("-" * 40)
print("Przykładowa pierwsza sekwencja wejściowa (X[0]):")
print(X[0])
print("-" * 40)
print(f"Pierwszy cel wyjściowy t[0] (średnia z X[0]): {t[0]}")
# Sprawdzenie: np.mean(X[0]) powinno być równe t[0, 0]
print(f"Weryfikacja: Średnia X[0] = {np.mean(X[0]):.4f}")
print("-" * 40)
print("Pierwsze 5 celów wyjściowych (t):")
print(t[:5])
```
