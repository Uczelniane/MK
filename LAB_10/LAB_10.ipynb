```python
import numpy as np

# Przykładowe dane do demonstracji algorytmu DMD.
n = 3
m = 10
r = 2

# Tworzenie losowych danych
np.random.seed(42)
X = np.random.rand(n, m)
X_prime = X @ np.random.rand(m, m) * 1.05

print(f"Rozmiar macierzy X: {X.shape}")
print(f"Rozmiar macierzy X': {X_prime.shape}")

X_dagger = np.linalg.pinv(X) # Obliczenie pseudoodwrotności X
A_approx = X_prime @ X_dagger
print("\n--- Podstawowe Obliczenie Macierzy A ---")
print(f"Rozmiar macierzy A_approx: {A_approx.shape}")

# Obliczenie SVD macierzy X z rzędem r
U, Sigma, Vh = np.linalg.svd(X, full_matrices=False)

# Redukcja do rzędu r
U_tilde = U[:, :r]
Sigma_tilde = np.diag(Sigma[:r])
V_tilde = Vh[:r, :].T

# Obliczenie odwrotności zredukowanej macierzy wartości osobliwych
Sigma_tilde_inv = np.linalg.inv(Sigma_tilde)

# Obliczenie zredukowanej macierzy A_tilde
A_tilde = U_tilde.T @ X_prime @ V_tilde @ Sigma_tilde_inv

print("\n--- Obliczenie Zredukowanej Macierzy A_tilde (DMD) ---")
print(f"Rozmiar macierzy A_tilde: {A_tilde.shape}")

# Obliczenie kompozycji spektralnej A_tilde
Lambda_diag, W = np.linalg.eig(A_tilde)

print("\n--- Kompozycja Spektralna A_tilde ---")
print(f"Wartości własne (Lambda): {Lambda_diag}")

# Rekonstrukcja niezredukowanych wektorów własnych (tryby DMD)
Phi = X_prime @ V_tilde @ Sigma_tilde_inv @ W

print("\n--- Rekonstrukcja Niezredukowanych Wektorów Własnych (Phi) ---")
print(f"Rozmiar macierzy Phi (Tryby DMD): {Phi.shape}")
```
